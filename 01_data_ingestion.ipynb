```python
#Load datasource
df = spark.read.csv("/Volumes/workspace/default/bank_marketing_raw/", header=True, sep=";", inferSchema=True)
display(df.limit(10))
```

```python
# Write the DataFrame to a Delta table in Databricks
df.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bank_marketing_bronze")
```

```python
# Save the ingested DataFrame as a Bronze Delta table
bronze_df = spark.table("workspace.default.bank_marketing_bronze")
#Check the number of records stored
print("Total Rows:", bronze_df.count())

#Check the number of columns
print("Total COlumns:", len(bronze_df.columns))

#Inspect data type of columns
bronze_df.printSchema()

#Preview first ten rows
display(bronze_df.limit(10))

bronze_df.explain()
```
